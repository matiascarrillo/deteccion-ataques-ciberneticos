Si analizamos el valor de los primeros 3 cuartiles de la columna 'sbytes' que representa cantidad de bytes enviados , encontramos una diferencia bastante importante con los valores maximos, por lo que calcularemos el rango intercuartilico para identificar outliers.

```python
#Primer cuartil
Q1 = df['sbytes'].quantile(0.25)
#Tercer cuartil
Q3 = df['sbytes'].quantile(0.75)

#Rango intercuartilico
IQR = Q3 - Q1

# Limite para detectar valores atipicos
limite_superior = Q3 + 1.5 * IQR

outliers = df[df['sbytes'] > limite_superior]

print("Límite superior:", limite_superior)
print("Cantidad de outliers:", len(outliers))
```

Encontramos que hay 22873 valores por encima del limite.

A continuación, averiguaremos cuantos de los outliers corresponden a trafico malicioso.

```python

outliers['label'].value_counts()
```

Se utilizó el método del rango intercuartílico (IQR) para identificar posibles valores atípicos en la variable sbytes, que representa la cantidad de bytes transmitidos desde el origen.

Se identificaron 22.873 registros que superan el límite superior de 3.374 bytes, equivalentes aproximadamente al 13,04 % del dataset.

De estos registros, 14.003 (61,2 %) corresponden a tráfico normal y 8.870 (38,8 %) a tráfico malicioso.

Estos resultados indican que los valores atípicos no necesariamente representan ataques informáticos. Se decidió conservarlos, ya que podrían contener información relevante para el análisis y la posterior construcción del modelo predictivo.

Con ésta informacion, procedemos a preparar las variables X e Y

```python
#Preparamos variable 'X'
X = df.drop(columns=['id','attack_cat', 'label'])
X.head()
```

```python
# Preparamos variable 'y'
y = df['label']
y.head()
```

Como anteriormente expusimos que existen muchas actividades con valores duplicados, creemos que es mejor mantener juntos esos valores en un solo conjunto, ya sea de entrenamiento o de prueba. Esto impedira que nuestro programa genere una capacidad de generalizacion y termine generando un overfitting.

```python
#Importamos GroupShuffleSplit que nos permite dividir los datos creando grupos
from sklearn.model_selection import GroupShuffleSplit

# Identificar grupos de conexiones idénticas
grupos = pd.util.hash_pandas_object(  #Crea identificadores de actividades identicas
    X, index=False  # sin tener en cuenta el indice
)

# Dividir sin repartir grupos entre ambos conjuntos, todos los duplicados
# queremos dejarlos juntos
division = GroupShuffleSplit(
    n_splits=1, #Cantidad de divisiones a realizar
    test_size=0.20, #tamaño de prueba
    random_state=42 #Fijar siempre mismos valores
)

 #next solicita la division de los datos y se lo facilita a train_idx y a test_idx
train_idx, test_idx = next(
    division.split(X, y, groups=grupos)
    #Decide que posiciones corresponden a entrenamiento y cuales a prueba,
    #mantieniene juntas las conexiones identicas
)
X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

```python
#Corroboramos que el porcentaje de ataques y de conexiones normales sea equitativa
print("Entrenamiento:")
print(y_train.value_counts(normalize=True) * 100)# Devuelve el porcentaje y no las cantidades

print("\nPrueba: ")
print(y_test.value_counts(normalize=True) * 100)
```

Como el porcentaje se encuentra equilibrado, procedemos a continuar preparando el contenido de las columnas para que el programa pueda analizar los datos de una manera mas comoda

```python
#Seleccionamos columnas del tipo 'object'
X.select_dtypes(include=['object']).columns
```

Vamos a proceder a preparar los datos para entrenar nuestro arbol de decision, para ello deberemos convertir nuestras columnas de tipo 'object' a un tipo numerico para que Scikit learn pueda trabajar con ellos.

```python
X_train = pd.get_dummies( #Transforma la columna de texto en 0 y 1
    X_train,
    columns=['proto', 'service', 'state'], #Indicamos las columnas a transformar
    dtype=int #Nos permite utilizar numeros enteros
)

X_test = pd.get_dummies(
    X_test,
    columns=['proto', 'service', 'state'],
    dtype=int
)

X_test = X_test.reindex(  # Igualamos las columnas de X_test con las de X_train
    columns=X_train.columns,  #Utiliza las columnas de X_train como referencia
    fill_value=0  #Completa con 0 las columnas que falten
)

print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)
```

Creacion del arbol de decision y su entrenamiento


```python
from sklearn.tree import DecisionTreeClassifier

modelo = DecisionTreeClassifier(
    max_depth=5,  #Limita profundidad del arbol a 5 niveles
    random_state=42 #Reproduce siempre los mismos resultados
)

#Entrenamos modelo
modelo.fit(X_train, y_train)
```

```python
#Realizamos las predicciones
y_pred = modelo.predict(X_test)

print(y_pred[:10])
```

Generamos la matriz de confusion, que nos permite comparar las predicciones del modelo con los resultados reales.

```python
from sklearn.metrics import confusion_matrix

matriz = confusion_matrix(y_test, y_pred)

print(matriz)
```

Nuestro arbol interpreto:

VN = 9041 --> Act normales bien clasificadas

FP = 2082 --> Act normales marcadas como ataques

FN = 154 --> Ataques no detectados

VP = 23919 --> Ataques detectados correctamente

A continuación, calcularemos las metricas para analizar que tan bueno es nuestro arbol

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

#Accuracy: Mide el porcentaje de predicciones correctas
print("Accuracy: %", accuracy_score(y_test, y_pred)*100)
#Precision: De todas las conexiones marcadas como ataque, que porcentaje lo era?
print("Precision: %", precision_score(y_test, y_pred)*100)
#Recall: De todas los ataques reales, que porcentaje se detecto?
print("Recall: %", recall_score(y_test, y_pred)*100)
#F1-score = Equilibrio entre precision y recall
print("F1-score: %", f1_score(y_test, y_pred)*100)
```

Accuracy (93,65 %): el modelo clasificó correctamente aproximadamente 94 de cada 100 conexiones.



Precision (91,99 %): de cada 100 conexiones que identificó como ataques, aproximadamente 92 realmente lo eran.



Recall (99,36 %): de cada 100 ataques reales, detectó aproximadamente 99.



F1-score (95,53 %): muestra un buen equilibrio entre Precision y Recall.

```python
print("Accuracy entrenamiento: %", modelo.score(X_train, y_train))
print("Accuracy prueba: %", modelo.score(X_test, y_test))
```

Entrenamos un árbol de decisión con profundidad máxima de 5 para controlar su complejidad y reducir el riesgo de sobreajuste. Obtuvimos un Accuracy del 93,50 % en entrenamiento y del 93,65 % en prueba. Como los resultados son muy similares, no encontramos evidencia importante de sobreajuste en esta comparación.

Conclusión:

El árbol logró identifficar el 99.36% de los ataques en el conjunto de prueba, con un Accuracy del 93.65%. Sin embargo, generó 2082 falsas alertas y dejo pasar 154 ataques.
